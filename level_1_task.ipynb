```python
import pandas as pd 

# read the csv file to find top rows of the dataset1.
df = pd.read_csv("Dataset1.csv")

print(" ---- top data columns list ----")
print(df.columns.tolist())

print(" ---- top 5 rows ----")
print(df.head(5))
```

```python
total_records, total_attributes = df.shape
print(f"Total records(Rows): {total_records}")
print(f"Total_attributes(columns): {total_attributes}")

#df.info(): Ye method pure dataset ka x-ray nikalta hai—har column ka naam, usme kitni non-empty (non-null) values hain, aur data type (int64, object/string, float64) show karta hai.
df.info()
```

```python
df_sorted = df.sort_values(by=['Train_No', 'SN'])
```

```python
train_terminals = df_sorted.groupby('Train_No').agg(starting_station = ('Station_Name', 'first'), ending_station =('Station_Name', 'last')).reset_index()
```

```python
train_terminals.head(10)
```

```python
#Har train number ke group me kitne station entries aayi hain, unhe count krenge.
train_stops = df.groupby('Train_No').count().reset_index()

#column ka name change kar denga (Station_Name -> Total_station)
train_stops.rename(columns={'Station_Name': 'Total_stops'}, inplace = True)

#records ko total stops ke hisab se descending order me arrange karenga ya sort karenga 
train_stops = train_stops.sort_values(by= 'Total_stops', ascending=False).reset_index(drop=True)

```

```python
train_stops.head(10)
```

```python
# train stops ki values me se sabse jada stops vala record return karenga
max_stops = train_stops['Total_stops'].max()

#minimun vali record return karenga 
min_stops = train_stops['Total_stops'].min()

#Ye check karta hai har row ke liye ki condition True hai ya False. Fir bahar wale train_stops[...] me sirf wahi rows aati hain jahan condition True ho.
trains_with_max_stops = train_stops[train_stops['Total_stops'] == max_stops]
trains_with_min_stops = train_stops[train_stops['Total_stops'] == min_stops]
```
